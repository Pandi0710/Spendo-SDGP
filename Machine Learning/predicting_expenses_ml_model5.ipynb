```python
#Importing pandas library
import pandas as pd
#Loading data into a DataFrame
modified_data=pd.read_csv('/content/drive/MyDrive/SDGP Implementation/modified_data (1).csv')
```

```python
modified_data.shape
```

```python
x5 = modified_data[['Income']]
y5 = modified_data['Education']
```

```python
x5
```

```python
y5
```

```python
from sklearn.model_selection import train_test_split
x5_train, x5_test, y5_train, y5_test = train_test_split(x5, y5, test_size=0.3, random_state=42)
```

```python
x5_train
```

```python
len(x5_test)
```

```python
from sklearn.linear_model import LinearRegression
clf5 = LinearRegression()
clf5.fit(x5_train, y5_train)
```

```python
clf5.predict(x5_test)
```

```python
y5_test
```

```python
clf5.score(x5_test, y5_test)
```

```python
clf5.coef_
```

```python
clf5.intercept_
```

```python
clf5.predict([[10000]])
```
