```python
#Importing pandas library
import pandas as pd
#Loading data into a DataFrame
modified_data=pd.read_csv('/content/drive/MyDrive/SDGP Implementation/modified_data (1).csv')
```

```python
modified_data.shape
```

```python
x3 = modified_data[['Income']]
y3 = modified_data['Eating_Out']
```

```python
x3
```

```python
y3
```

```python
from sklearn.model_selection import train_test_split
x3_train, x3_test, y3_train, y3_test = train_test_split(x3, y3, test_size=0.3, random_state=42)
```

```python
x3_train
```

```python
len(x3_test)
```

```python
from sklearn.linear_model import LinearRegression
clf3 = LinearRegression()
clf3.fit(x3_train, y3_train)
```

```python
clf3.predict(x3_test)
```

```python
y3_test
```

```python
clf3.score(x3_test, y3_test)
```

```python
clf3.coef_
```

```python
clf3.intercept_
```

```python
clf3.predict([[10000]])
```
