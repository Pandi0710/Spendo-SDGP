```python
#Importing pandas library
import pandas as pd
#Loading data into a DataFrame
modified_data=pd.read_csv('/content/drive/MyDrive/SDGP Implementation/modified_data (1).csv')
```

```python
modified_data.shape
```

```python
x4 = modified_data[['Income']]
y4 = modified_data['Healthcare']
```

```python
x4
```

```python
y4
```

```python
from sklearn.model_selection import train_test_split
x4_train, x4_test, y4_train, y4_test = train_test_split(x4, y4, test_size=0.3, random_state=42)
```

```python
x4_train
```

```python
len(x4_test)
```

```python
from sklearn.linear_model import LinearRegression
clf4 = LinearRegression()
clf4.fit(x4_train, y4_train)
```

```python
clf4.predict(x4_test)
```

```python
y4_test
```

```python
clf4.score(x4_test, y4_test)
```

```python
clf4.coef_
```

```python
clf4.intercept_
```

```python
clf4.predict([[10000]])
```
