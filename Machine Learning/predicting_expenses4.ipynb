```python
#Importing pandas library
import pandas as pd
#Loading data into a DataFrame
cleaned_data2=pd.read_csv('/content/drive/MyDrive/SDGP Implementation/remove_out2 (1).csv')
```

```python
cleaned_data2.shape
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.scatter(cleaned_data2['Income'],cleaned_data2['Eating_Out'])
```

```python
def find_outliers_IQR(cleaned_data2):
  q1=cleaned_data2.quantile(0.25)
  q3=cleaned_data2.quantile(0.75)
  IQR=q3-q1
  outliers = cleaned_data2[((cleaned_data2<(q1-1.5*IQR))|(cleaned_data2>(q3+1.5*IQR)))]
  return outliers
```

```python
outliers = find_outliers_IQR(cleaned_data2['Eating_Out'])
print("number of outliers: "+ str(len(outliers)))
outliers
```

```python
def remove_outliers_IQR(cleaned_data2, column):
    Q1 = cleaned_data2[column].quantile(0.25)
    Q3 = cleaned_data2[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return cleaned_data2[(cleaned_data2[column] >= lower_bound) & (cleaned_data2[column] <= upper_bound)]
```

```python
outliers = find_outliers_IQR(cleaned_data2['Eating_Out'])
print("Number of outliers before removal:", len(outliers))
```

```python
cleaned_data3 = remove_outliers_IQR(cleaned_data2, 'Eating_Out')
```

```python
print("Number of records after outlier removal:", len(cleaned_data3))
```

```python
# Save the cleaned dataset
cleaned_data3.to_csv('remove_out3.csv', index=False)

# Download the new file
from google.colab import files
files.download('remove_out3.csv')

# Display the first few rows
cleaned_data3.head()
```
